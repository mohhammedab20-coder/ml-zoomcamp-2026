```python
import pandas as pd

url = "https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv"

df = pd.read_csv(url)

# Q4: عدد الأعمدة التي تحتوي على قيم مفقودة
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nNumber of columns with missing values:")
print((missing_values > 0).sum())
```

```python
print(df.columns.tolist())
```

```python
asia = df[df['origin'] == 'Asia']

print(asia['fuel_efficiency_mpg'].max())
```

```python
# Q6

# Median before filling missing values
old_median = df['horsepower'].median()

# Find the most frequent horsepower value
most_frequent = df['horsepower'].mode()[0]

# Fill missing horsepower values with the mode
df['horsepower'] = df['horsepower'].fillna(most_frequent)

# Median after filling missing values
new_median = df['horsepower'].median()

print("Old median:", old_median)
print("Most frequent value:", most_frequent)
print("New median:", new_median)

if new_median > old_median:
    print("Answer: Yes, it increased")
elif new_median < old_median:
    print("Answer: No, it decreased")
else:
    print("Answer: No, it did not change")
```

```python
import numpy as np

asia = df[df['origin'] == 'Asia']

X = asia[['vehicle_weight', 'model_year']].head(7).to_numpy()

print("X:")
print(X)

XTX = X.T @ X

print("XTX:")
print(XTX)

XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y

print("w:")
print(w)

print("Sum of w:")
print(w.sum())
```
